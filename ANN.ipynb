```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("students_dropout_dataset.csv")
data
```

```python
data.isnull().sum()
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
data.columns = data.columns.str.strip()
```

```python
X = data[['attendance_pct','assignment_completion_pct','avg_quiz_score','logins_per_week','avg_learning_hours_per_week']].values
X
```

```python
y = data['dropped_out'].values
y
```

```python
model = Sequential()
model.add(Dense(8, input_dim=5, activation='relu'))
model.add(Dense(2, activation='sigmoid'))
```

```python
model.compile(loss='sparse_categorical_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])
```

```python
model.fit(X, y, epochs=80)
```

```python
att_per = float(input("Enter attendance percentage: "))

assignment_com_pct = float(input("Enter assignment completion percentage: "))

avg_quiz_score = float(input("Enter average quiz score: "))

logins_per_week = float(input("Enter number of logins per week: "))

learning_hours = float(input("Enter average learning hours per week: "))
```

```python
test = np.array([[att_per,assignment_com_pct,avg_quiz_score,logins_per_week,learning_hours]])
pred = model.predict(test)
```

```python
dropout_probability = pred[0][0]

if dropout_probability >= 0.5:
    print("Prediction: High Dropout Risk")
else:
    print("Prediction: Low Dropout Risk")
```
